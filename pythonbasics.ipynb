## Match Case Statement 

```python
def http_status(code):
    if code == 200:
        return "OK"
    elif code == 400:
        return "Bad Request"
    elif code == 404:
        return "Not Found"
    elif code == 500:
        return "Internal Server Error"
    else:
        return "Unknown Status"

print(http_status(200))
```

# File Handling 

```python
# Allow to handle file read, write and manipulation
file = open("example.txt", "r")
content = file.read()
print(content)
file.close()
```

```python
file = open("example.txt", "r")
content1 = file.readline()
print(content1)
file.close()
```

```python
#Read all lines as list
file = open("example.txt", "r")
lines = file.readlines()
print(lines)
file.close()
```

## Write 

```python
file = open("example.txt", "w")
file.write("Hello, World!")
file.close()
```

```python
lines = ["Hello\n", "Welcome to python\n", "File Handling\n"]
file = open("example.txt", "w")
file.writelines(lines)
file.close()
```

## Append

```python
#in append mode , we can add extra content without removing previous unlike normal write
file = open("example.txt", "a")
file.write("\nThis is an additiona line")
file.close()
```

```python
file = open("example.txt", "r")
content = file.read()
print(content)
file.close()
```

# file handling using with statement 

```python
with open("example.txt", "r") as file:
    content = file.read()
    print(content)
```

# Checking if file exist or not 

```python
# Use the os module to check if a file exists before opening it.
import os 

if os.path.exists("example.txt"):
    print("File exists!")
else:
    print("File not found")
```

## Deleting a file

```python
import os

if os.path.exists("example.txt"):
    os.remove("example.txt")
    print("File delted")
else:
    print("File not found")
```

# JSON

```python
# JSON (JavaScript Object Notation) is a lightweight data format used for data
# exchange between servers and applications. It is widely used in APIs, web
# applications, and configurations.
```

```python
import json
```

```python
#converting python object to JSON string 
import json

data = {"name": "Alice", "age" : 25, "City" : "New York"}
json_string = json.dumps(data)
print(json_string)
print(type(json_string))
```

```python
#converting python object to file
with open("data.json", "w") as file:
    json.dump(data, file)
```

```python
#Convert python object to json
json_data = '{"name":"Alice", "age":25, "City": "New York"}'
print(type(json_data))

#converting json to python 
python_obj = json.loads(json_data)
print(python_obj)
print(type(python_obj))
```
