```python
from ultralytics import YOLO

# Load pretrained YOLOv8s model
model = YOLO("yolov8s.pt")

# Train
model.train(
    data="/kaggle/input/datasets/garambharadhi/major-project/data.yaml",  # <-- apna exact path yahan do
    epochs=120,          # better learning
    imgsz=832,           # small drone/bird ke liye better
    batch=8,
    device=0,            # GPU
    workers=2,
    cache=True,
    pretrained=True,     # fast + better accuracy
    patience=30,         # early stopping

    # augmentations
    mosaic=1.0,
    mixup=0.15,
    fliplr=0.5,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,

    project="/kaggle/working/results",
    name="train_results"
)
```
